## Streaming

LLM streaming means the model sends generated text token by token (or chunk by chunk) instead of waiting for the full response to complete.

```python
for chunk in llm.stream("Write a 200 para on AI"):
    print(chunk.text,end="|",flush=True)
```

## Batching

Batching in LLMs means processing multiple requests together in a single model execution instead of handling one request at a time.

```python
responses=llm.batch([
    "Hello how are you",
    "I am sanskar",
    "Who are you"
])

for response in responses:
    print(response)
```

```python
responses=llm.batch([
    "Hello how are you",
    "I am sanskar",
    "Who are you"
],
config={
    "max_concurrency":5 # Limit to 5 parallel calls
})

for response in responses:
    print(response)
```
